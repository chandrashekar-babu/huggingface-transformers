### Setup environment for HuggingFace

#### Setup using Anaconda Python

First, download and install Anaconda Python from http://www.anaconda.com/ (miniconda is preferred)

For Windows (x86 environments):
  - https://anaconda.com/api/installers/Miniconda3-latest-Windows-x86_64.exe

For MacOS (Apple Silicon M3/M4+):
  - https://anaconda.com/api/installers/Miniconda3-latest-MacOSX-arm64.pkg

For Linux / WSL (x86):
  - https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh

On Windows (without WSL) and MacOS, just run the downloaded executable, follow the instructions (ensure installation is for Local user, not everyone)

On Linux / WSL (on Windows), open the terminal shell, locate the downloaded file and run the following command:
```bash
   chmod +x Miniconda3-latest-Linux-x86_64.sh  # Ensure the right path to the file
   ./Miniconda3-latest-Linux-x86_64.sh # Follow instructions, install in user's home folder
```

Now, to open Anaconda environment on Windows (without WSL): Run "Anaconda Command Prompt" from the start menu (search if needed). You should see a windows command prompt in a terminal that looks like this: ```(base) C:\Users\john>```

For Linux, WSL, MacOS, open up a new terminal (restart the shell if needed) and verify that python comes from anaconda's environment by typing ```which python``` which should show something like ```/home/john/anaconda3/bin/python```

### Using the ```pipeline()``` from the ```transformers``` library

### Using the low-level API for more tasks

````python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

model_name = "openai-community/gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

model_inputs = tokenizer("This distance between the Earth and the Moon in kilometers is", return_tensors="pt")
generated_ids = model.generate(**model_inputs, max_new_tokens=20)
generated_text = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
print(generated_text)
````
